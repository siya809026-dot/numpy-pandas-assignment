# 7) ravel() vs flatten() — view vs copy

# ravel() — 3 implementations

```python
# Implementation 1: Classroom seating numbers (2x2 grid)
base1 = np.array([[1, 2], [3, 4]])
r1 = base1.ravel()
r1[0] = 999
print(base1)   # changed — ravel() is a view, linked to original
```

```python
# Implementation 2: Product price grid (2 products x 2 warehouses)
base2 = np.array([[500, 750], [300, 620]])
r2 = base2.ravel()
r2[0] = 9999
print(base2)   # changed — same memory as original
```

```python
# Implementation 3: Temperature readings (2 days x 2 time slots)
base3 = np.array([[22.5, 28.0], [24.1, 29.3]])
r3 = base3.ravel()
r3[0] = 0.0
print(base3)   # changed — ravel() reflects back on original
```

# flatten() — 3 implementations

```python

base4 = np.array([[1, 2], [3, 4]])
f1 = base4.flatten()
f1[0] = 111
print(base4)
```

```python
scores = np.array([[85, 90], [78, 92]])   # original marks, safe

temp = scores.flatten()   
temp[0] = 0                

print(scores)   
print(temp)     
```

```python

marks = np.array([[85, 78], [92, 88], [60, 70]])


all_marks = marks.flatten()
highest = np.sort(all_marks)[-1]   

print(highest)  
print(marks)     
```

# 8) Boolean masking with & (AND)

```python

data1 = np.array([45, 55, 60, 90, 35, 72, 88, 50])
mask1 = (data1 > 50) & (data1 < 80)
print(data1[mask1])
```

```python

data2 = np.array([4, 15, 22, 7, 30, 9, 16])
mask2 = (data2 % 2 == 0) & (data2 > 10)
print(data2[mask2])
```

```python

data3 = np.array([35, 42, 58, 95, 61, 33, 78])
mask3 = (data3 >= 33) & (data3 != 95)   

print(data3[mask3])
```

# 12) Boolean masking with ~ (NOT / inverse)

```python
marks = np.array([35, 72, 28, 90, 45, 20])

mask = (marks >= 40)

print(marks[mask])     # Passed
print(marks[~mask])    # Failed
```

```python
prices = np.array([500, 1200, 800, 2500, 1500, 300])

mask = (prices >= 1000)

print(prices[mask])     # ₹1000 or more
print(prices[~mask])    # Below ₹1000
```

```python
temperature = np.array([25, 32, 38, 29, 41, 27])

mask = (temperature >= 35)

print(temperature[mask])     # Hot
print(temperature[~mask])    # Not hot
```

# 13) Range masking with & (between two values)


```python
numbers = np.array([10, 15, 20, 25, 30, 35, 40])

mask = (numbers % 2 == 0) & (numbers % 5 == 0)

print(numbers[mask])
```

```python
ages = np.array([18, 20, 18, 22, 21, 18])

mask = (ages == 18) & (ages % 2 == 0)

print(ages[mask])
```

```python
numbers = np.array([12, 15, 20, 30, 45, 50, 60])

mask = (numbers % 3 == 0) & (numbers % 5 == 0)

print(numbers[mask])
```

# 14) np.select() — single condition


```python
age = np.array([12, 18, 15, 21, 16])

print(np.select([age >= 18], ["Adult"], default="Minor"))
```

```python
attendance = np.array([65, 80, 75, 90, 55])

print(np.select([attendance >= 75], ["Eligible"], default="Not Eligible"))
```

```python
numbers = np.array([10, -5, 8, -2, 15])

print(np.select([numbers >= 0], ["Positive"], default="Negative"))
```

# 15) np.select() — multiple conditions with default


```python
marks = np.array([95, 82, 76, 68, 55, 43, 31, 89, 72, 60, 97, 25])

cond = [
    (marks >= 80) & (marks <= 100),
    (marks >= 60) & (marks < 80)
]

choice = ["A Grade", "B Grade"]

print(np.select(cond, choice, default="C Grade"))
```

```python
experience = np.array([0, 1, 2, 3, 4, 5, 6, 8, 10, 2, 7, 12])

cond = [
    (experience >= 3) & (experience <= 5),
    (experience > 5) & (experience <= 10)
]

choice = ["Mid-Level", "Senior"]

print(np.select(cond, choice, default="Fresher"))
```

```python
orders = np.array([
    350, 600, 850, 1200, 1500, 1800,
    2200, 750, 950, 1300, 2500, 450
])

cond = [
    (orders >= 500) & (orders <= 999),
    (orders >= 1000) & (orders <= 1999)
]

choice = ["5% Discount", "10% Discount"]

print(np.select(cond, choice, default="No Discount"))
```
