# Apply a Function to Data

```python
student_data = {
    "Name": ["Siya", "Aman", "Riya", "Karan"],
    "Marks": [85, 92, 78, 88]
}

students7 = pd.DataFrame(student_data)

def grade(marks):
    if marks >= 90:
        return "A"
    elif marks >= 80:
        return "B"
    else:
        return "C"

students7["Grade"] = students7["Marks"].apply(grade)

print(students7)
```

```python
employee_data = {
    "Employee": ["Aman", "Riya", "Karan", "Neha"],
    "Monthly_Salary": [45000, 40000, 52000, 48000]
}

employees7 = pd.DataFrame(employee_data)

def annual_salary(salary):
    return salary * 12

employees7["Annual_Salary"] = employees7["Monthly_Salary"].apply(annual_salary)

print(employees7)
```

```python
product_data = {
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [55000, 800, 1500, 12000]
}

products7 = pd.DataFrame(product_data)

def discount(price):
    return price * 0.90

products7["Discounted_Price"] = products7["Price"].apply(discount)

print(products7)
```

# Group Data for Analysis

```python
student_data = {
    "Name": ["Siya", "Aman", "Riya", "Karan", "Neha", "Rohit"],
    "Class": ["A", "A", "B", "B", "A", "B"],
    "Marks": [85, 92, 78, 88, 90, 75]
}

students8 = pd.DataFrame(student_data)

result = students8.groupby("Class")["Marks"].mean()

print(result)
```

```python
employee_data = {
    "Employee": ["Aman", "Riya", "Karan", "Neha", "Rohit", "Priya"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "Finance"],
    "Salary": [45000, 40000, 52000, 48000, 42000, 55000]
}

employees8 = pd.DataFrame(employee_data)

result = employees8.groupby("Department")["Salary"].mean()

print(result)
```

```python
sales_data = {
    "Product": ["Laptop", "Mouse", "Laptop", "Keyboard", "Mouse", "Laptop"],
    "Sales": [55000, 800, 60000, 1500, 1200, 50000]
}

sales8 = pd.DataFrame(sales_data)

result = sales8.groupby("Product")["Sales"].sum()

print(result)
```

# Data ko label/condition ke basis par select karna

```python
students = pd.DataFrame({
    "Name": ["Siya", "Aman", "Riya", "Karan"],
    "Marks": [85, 92, 78, 88],
    "Attendance": [90, 95, 82, 89]
})

result = students.loc[students["Marks"] >= 85]

print(result)
```

```python
employees = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Department": ["IT", "HR", "IT", "Finance"],
    "Salary": [50000, 45000, 60000, 55000]
})

result = employees.loc[:, ["Name", "Salary"]]

print(result)
```

```python
products = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "Price": [55000, 800, 1500, 12000],
    "Stock": [10, 50, 30, 15]
})

result = products.loc[products["Price"] > 1000, ["Product", "Price"]]

print(result)
```

# Multiple DataFrames ko combine karna


```python
class_a = pd.DataFrame({
    "Name": ["Siya", "Aman"],
    "Marks": [85, 92]
})

class_b = pd.DataFrame({
    "Name": ["Riya", "Karan"],
    "Marks": [78, 88]
})

result = pd.concat([class_a, class_b], ignore_index=True)

print(result)
```

```python
january = pd.DataFrame({
    "Product": ["Laptop", "Mouse"],
    "Sales": [10, 25]
})

february = pd.DataFrame({
    "Product": ["Laptop", "Mouse"],
    "Sales": [15, 30]
})

result = pd.concat([january, february], ignore_index=True)

print(result)
```

```python
it = pd.DataFrame({
    "Name": ["Aman", "Riya"],
    "Department": ["IT", "IT"],
    "Salary": [50000, 55000]
})

hr = pd.DataFrame({
    "Name": ["Karan", "Neha"],
    "Department": ["HR", "HR"],
    "Salary": [45000, 48000]
})

result = pd.concat([it, hr], ignore_index=True)

print(result)
```
